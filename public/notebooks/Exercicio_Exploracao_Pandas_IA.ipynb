## 1. Preparação

Vamos utilizar um pequeno conjunto de dados de vendas. Ele contém informações sobre produtos, categorias, cidades, quantidades vendidas, preços e avaliações.

O conjunto foi criado especialmente para o exercício e possui alguns valores ausentes para que possamos praticar a identificação de problemas nos dados.

```python
import pandas as pd

print("Pandas carregado com sucesso!")
```

## 2. Carregando os dados

O código abaixo cria um pequeno arquivo CSV e depois o carrega usando `pd.read_csv()`.

Observe que, em um projeto real, o arquivo poderia vir de uma base de dados, de um sistema ou de uma fonte pública já em CSV ou em outros formatos.

```python
dados = {
    "produto": ["Notebook", "Mouse", "Teclado", "Monitor", "Mouse", "Notebook", "Headset", "Teclado", "Monitor", "Mouse", "Headset", "Notebook", "Teclado", "Monitor", "Mouse", "Headset", "Notebook", "Teclado", "Monitor", "Mouse"],
    "categoria": ["Computador", "Periférico", "Periférico", "Monitor", "Periférico", "Computador", "Áudio", "Periférico", "Monitor", "Periférico", "Áudio", "Computador", "Periférico", "Monitor", "Periférico", "Áudio", "Computador", "Periférico", "Monitor", "Periférico"],
    "cidade": ["Recife", "Olinda", "Recife", "Jaboatão", "Recife", "Olinda", "Recife", "Jaboatão", "Recife", "Olinda", "Recife", "Recife", "Olinda", "Jaboatão", "Recife", "Olinda", "Jaboatão", "Recife", "Olinda", "Recife"],
    "quantidade": [2, 5, 3, 1, 4, 1, 6, 2, 2, 7, 3, 1, 4, 1, 5, 2, 2, 3, 1, 6],
    "preco": [3500.0, 80.0, 150.0, 1200.0, 80.0, 3500.0, 250.0, 150.0, 1200.0, 80.0, 250.0, 3500.0, 150.0, 1200.0, 80.0, 250.0, 3500.0, 150.0, 1200.0, 80.0],
    "avaliacao": [4.8, 4.2, 4.5, 4.7, None, 4.9, 4.1, 4.4, 4.6, 4.0, 4.3, None, 4.5, 4.8, 4.1, 4.2, 4.9, None, 4.7, 4.0]
}

df_inicial = pd.DataFrame(dados)
df_inicial.to_csv("vendas.csv", index=False)

df = pd.read_csv("vendas.csv")
df = df.astype({
    "produto": "string",
    "categoria": "string",
    "cidade": "string"
})

df.head()
```
